```python
from pathlib import Path
import xml.etree.ElementTree as ET
from collections import Counter

ECUST_ROOT = Path(
    "/Users/amirsadra213/Desktop/MFT Class/Deep Learning/Food Detection/ECUSTFD-resized--master"
)

ANNOT_DIR = ECUST_ROOT / "Annotations"

class_counts = Counter()
object_counts = Counter()

xml_files = list(ANNOT_DIR.glob("*.xml"))

print("XML files:", len(xml_files))

for xml_file in xml_files:
    tree = ET.parse(xml_file)
    root = tree.getroot()

    objects = root.findall("object")
    object_counts[len(objects)] += 1

    for obj in objects:
        class_name = obj.find("name").text.strip()
        class_counts[class_name] += 1

print("\nClasses:")
for cls, count in sorted(class_counts.items()):
    print(f"{cls:20s} {count}")

print("\nNumber of objects per image:")
for n, count in sorted(object_counts.items()):
    print(f"{n} object(s): {count} images")
```

```python
sample_xml = xml_files[0]

tree = ET.parse(sample_xml)
root = tree.getroot()

print("File:", sample_xml.name)

print("Filename:", root.findtext("filename"))

size = root.find("size")
if size is not None:
    print(
        "Image size:",
        size.findtext("width"),
        "x",
        size.findtext("height")
    )

for obj in root.findall("object"):
    print("\nClass:", obj.findtext("name"))

    bbox = obj.find("bndbox")
    if bbox is not None:
        print(
            "BBox:",
            bbox.findtext("xmin"),
            bbox.findtext("ymin"),
            bbox.findtext("xmax"),
            bbox.findtext("ymax")
        )
```

```python
from pathlib import Path
import re
from collections import defaultdict

image_dir = ECUST_ROOT / "JPEGImages"

SELECTED_ECUST_CLASSES = [
    "apple",
    "banana",
    "tomato",
    "lemon",
    "litchi",
    "mango",
    "orange",
    "peach",
    "pear",
    "plum",
    "qiwi"
]

samples_by_class = defaultdict(lambda: defaultdict(list))

for class_name in SELECTED_ECUST_CLASSES:
    for image_path in image_dir.glob(f"{class_name}*.JPG"):

        # Example:
        # tomato001S(8).JPG
        #           ↓
        # tomato001

        match = re.match(
            rf"({re.escape(class_name)}\d+)[ST]\(\d+\)",
            image_path.stem
        )

        if match:
            sample_id = match.group(1)
            samples_by_class[class_name][sample_id].append(image_path)

print(f"{'Class':<12} {'Samples':>8} {'Images':>8}")
print("-" * 30)

total_samples = 0
total_images = 0

for class_name in SELECTED_ECUST_CLASSES:

    samples = samples_by_class[class_name]

    n_samples = len(samples)
    n_images = sum(len(images) for images in samples.values())

    total_samples += n_samples
    total_images += n_images

    display_name = "kiwi" if class_name == "qiwi" else class_name

    print(f"{display_name:<12} {n_samples:>8} {n_images:>8}")

print("-" * 30)
print(f"{'TOTAL':<12} {total_samples:>8} {total_images:>8}")
```

```python
import re
from pathlib import Path
from collections import defaultdict

image_dir = ECUST_ROOT / "JPEGImages"

SELECTED_ECUST_CLASSES = [
    "apple",
    "banana",
    "tomato",
    "lemon",
    "litchi",
    "mango",
    "orange",
    "peach",
    "pear",
    "plum",
    "qiwi"
]

print(f"{'Class':<12} {'All':>6} {'Grouped':>8} {'Missing':>8}")
print("-" * 38)

all_missing = []

for class_name in SELECTED_ECUST_CLASSES:

    all_images = list(image_dir.glob(f"{class_name}*.JPG"))

    grouped_images = []

    for sample_images in samples_by_class[class_name].values():
        grouped_images.extend(sample_images)

    all_names = {p.name for p in all_images}
    grouped_names = {p.name for p in grouped_images}

    missing = sorted(all_names - grouped_names)

    display_name = "kiwi" if class_name == "qiwi" else class_name

    print(
        f"{display_name:<12}"
        f"{len(all_images):>6}"
        f"{len(grouped_images):>8}"
        f"{len(missing):>8}"
    )

    for name in missing:
        all_missing.append((class_name, name))

print("\nTotal missing:", len(all_missing))

print("\nFirst missing filenames:")
for cls, name in all_missing[:50]:
    print(f"{cls:<10} -> {name}")
```
